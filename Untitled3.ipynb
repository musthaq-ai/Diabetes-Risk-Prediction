```python
import pandas as pd

df = pd.read_csv('diabetes.csv')
print("Shape:", df.shape)
print("\nFirst 5 rows:")
df.head()
```

```python
# Basic info
print("Column names:", df.columns.tolist())
print("\nData types:\n", df.dtypes)
print("\nMissing values:\n", df.isnull().sum())
print("\nBasic statistics:")
df.describe()
```

```python
from sklearn.preprocessing import StandardScaler

# Some columns have 0 values which are actually missing (e.g., Glucose, BP can't be 0)
cols_with_zeros = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

# Replace 0s with column mean
for col in cols_with_zeros:
    df[col] = df[col].replace(0, df[col].mean())

# Split features and target
X = df.drop('Outcome', axis=1)
y = df['Outcome']

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Preprocessing done! ✅")
print("Features shape:", X_scaled.shape)
print("Target distribution:\n", y.value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

print("Train size:", X_train.shape[0])
print("Test size:", X_test.shape[0])

# Train Logistic Regression
lr = LogisticRegression()
lr.fit(X_train, y_train)

# Train Random Forest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

print("\nModels trained! ✅")
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import seaborn as sns
import matplotlib.pyplot as plt

# Predictions
lr_pred = lr.predict(X_test)
rf_pred = rf.predict(X_test)

# Accuracy
lr_acc = accuracy_score(y_test, lr_pred)
rf_acc = accuracy_score(y_test, rf_pred)

print(f"Logistic Regression Accuracy: {lr_acc*100:.2f}%")
print(f"Random Forest Accuracy:       {rf_acc*100:.2f}%")

# Classification Report
print("\nRandom Forest Report:")
print(classification_report(y_test, rf_pred))

# Confusion Matrix - Random Forest
cm = confusion_matrix(y_test, rf_pred)
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['No Diabetes','Diabetes'],
            yticklabels=['No Diabetes','Diabetes'])
plt.title('Random Forest - Confusion Matrix')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.tight_layout()
plt.show()

print("\nEvaluation done! ✅")
```

```python
import pickle

# Save best model (Random Forest)
with open('diabetes_model.pkl', 'wb') as f:
    pickle.dump(rf, f)

# Save scaler too
with open('scaler.pkl', 'wb') as f:
    pickle.dump(scaler, f)

print("Model saved! ✅")

# Quick test - predict one sample
sample = X_test[0].reshape(1, -1)
prediction = rf.predict(sample)
print(f"\nSample prediction: {'Diabetic' if prediction[0]==1 else 'Not Diabetic'}")
```
